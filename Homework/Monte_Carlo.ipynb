```python
import matplotlib.pyplot as plt
import numpy as np
import random
import math


#Define Monte Carlo Function. Input samples, output estimated pi.
def MonteCristoPi(Samples):
    inside = 0
    outside = 0
    for i in range(Samples):
        x = random.uniform(0.0, 10.0)
        y = random.uniform(0.0, 10.0)
        if ((x*x + y*y) <= 100):
            inside += 1
        else:
            outside += 1
    
    #inside/total samples times 4. 
    #Still works for r = 10 because r^2 is in the formula for both the square and the circle,
    #So the ratio circleSector/Square is the same: pi/4.
    pi = 4*(inside/Samples)
    return(pi)

t = []      #Define list axes for plotting. t = samples (on x-axis), error = on y-axis
error = []

#Outer loop iterates through different Monte Carlo sample amounts
#Inner loop averages results from outputs with the same number of samples. Creates a more consistent/less noisy graph.
for j in range(10, 1000, 20):
    piTemp = MonteCristoPi(j)
    errorTemp = abs(math.pi-piTemp)

    #Add vals to lists.
    t.append(j)    
    error.append(errorTemp)

plt.plot(t, error)



#Used AI to obtain a random sampler (ie. import random and random.uniform(float, float))
```
